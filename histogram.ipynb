# Count Occurrences of Each Patient ID
Count the occurrences of each patient ID using pandas.

```python
# Count Occurrences of Each Patient ID
patient_id_counts = df['Patient_ID'].value_counts()

# Display the counts
patient_id_counts.head()

# Plot the histogram
plt.figure(figsize=(100, 6))
patient_id_counts.plot(kind='bar')
plt.title('Occurrences of Each Patient ID')
plt.xlabel('Patient ID')
plt.ylabel('Count')
plt.show()
```
